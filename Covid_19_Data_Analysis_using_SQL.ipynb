## Q14 — HAVING

**Task:** Find countries whose total confirmed cases are greater than 50,000,000.

Write the SQL query in the next cell.

```python
%%sql

SELECT Country, SUM(ConfirmedCases) AS TotalCases
FROM Covid19_Data
GROUP BY Country
HAVING SUM(ConfirmedCases) > 50000000;
```

## Q15 — HAVING deaths

**Task:** Find countries whose total deaths are greater than 500,000.

Write the SQL query in the next cell.

```python
%%sql
select Country, (sum(Deaths)+sum(NewDeaths)) as Total_Deaths from covid19_Data 
group by COuntry having (sum(Deaths)+sum(NewDeaths)) >500000;
```

## Q16 — CASE classification

**Task:** Create a CaseLevel column: >=100000 = Very High, >=50000 = High, >=10000 = Medium, otherwise Low.

Write the SQL query in the next cell.

```python
%%sql

select Newcases,
case 
when Newcases>=100000 then 'Very High'
when Newcases>=50000 then 'High'
when Newcases>=10000 then 'Medium'
else 'Low'
End as Caselevel
from Covid19_data;
```

## Q18 — Death rate

## Q19 — Top country

**Task:** Find the country with the highest total confirmed cases. Use GROUP BY, ORDER BY and TOP.

Write the SQL query in the next cell.

```python
%%sql

select top 1 sum(ConfirmedCases) as Total_Highest_Confirmed_Case, country from covid19_Data
group by country order by Total_Highest_Confirmed_Case Desc;
```

## Q20 — Country summary

**Task:** For each country display Total Confirmed Cases, Total Deaths, Total Recovered Cases and Death Rate. Sort by Total Confirmed Cases descending.

Write the SQL query in the next cell.

```python
%%sql
select country, sum(ConfirmedCases) as Total_Confirmed_Cases ,
(sum(Deaths)+sum(NewDeaths)) as TotalDeaths, 
sum(RecoveredCases) as Total_Recovered_Cases
, (sum(Deaths)*100/sum(ConfirmedCases))as Death_Rate 
from Covid19_Data 
group by 
country order by Total_Confirmed_Cases Desc;
```
